```python
numbers = [-4, -3, -2, -1, 0, 2, 4, 6]

filtered = [num for num in numbers if num <= 0]
print(filtered)
```

```python
list_of_lists = [[[1, 2, 3]], [[4, 5, 6]], [[7, 8, 9]]]

flattened = [num for first in list_of_lists for second in first for num in second]
print(flattened)
```

```python
result = [(i, 1, i, i**2, i**3, i**4, i**5,) for i in range(11)]
print(result)
```

```python
countries = [[('Finland', 'Helsinki')], [('Sweden', 'Stockholm')], [('Norway', 'Oslo')]]

result = [[country.upper(), country[:3].upper(), city.upper()] for sub in countries for country, city in sub]
print(result)
```

```python
countries = [[('Finland', 'Helsinki')], [('Sweden', 'Stockholm')], [('Norway', 'Oslo')]]

result = [{'country': country.upper(), 'city': city.upper()} for sub in countries for country, city in sub]
print(result)
```

```python
names = [[('Asabeneh', 'Yetayeh')], [('David', 'Smith')], [('Donald', 'Trump')], [('Bill', 'Gates')]]

result = [first + ' ' + last for sub in names for first, last in sub]
print(result)
```

```python
slope = lambda x1, y1, x2, y2: (y2 - y1) / (x2 - x1)

y_intercept = lambda x1, y1, x2, y2: y1 - slope(x1, y1, x2, y2) * x1

print("Slope:", slope(1, 2, 3, 6))
print("Y-intercept:", y_intercept(1, 2, 3, 6))
```
