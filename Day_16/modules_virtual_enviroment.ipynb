```python
import random
import string

def random_user_id():
    chars = string.ascii_letters + string.digits
    user_id = ''
    for _ in range(6):
        user_id += random.choice(chars)
    return user_id

print(random_user_id())
```

```python
import random
import string

def user_id_gen_by_user():
    num_chars = int(input("Enter number of characters: "))
    num_ids = int(input("Enter number of IDs to generate: "))
    chars = string.ascii_letters + string.digits

    for _ in range(num_ids):
        user_id = ''.join(random.choices(chars, k=num_chars))
        print('#' + user_id)

user_id_gen_by_user()
```

```python
import random

def list_of_hexa_colors(n):
    colors = []
    hex_chars = '0123456789abcdef'
    for _ in range(n):
        color = '#' + ''.join(random.choices(hex_chars, k=6))
        colors.append(color)
    return colors

print(list_of_hexa_colors(3))
```

```python
import random

def list_of_rgb_colors(n):
    colors = []
    for _ in range(n):
        r = random.randint(0, 255)
        g = random.randint(0, 255)
        b = random.randint(0, 255)
        colors.append(f'rgb({r}, {g}, {b})')
    return colors

print(list_of_rgb_colors(3))
```

```python
import random

def generate_colors(type, n):
    if type == 'hexa':
        hex_chars = '0123456789abcdef'
        return ['#' + ''.join(random.choices(hex_chars, k=6)) for _ in range(n)]

    elif type == 'rgb':
        return [f'rgb({random.randint(0,255)}, {random.randint(0,255)}, {random.randint(0,255)})' for _ in range(n)]

print(generate_colors('hexa', 3))
print(generate_colors('hexa', 1))
print(generate_colors('rgb', 3))
print(generate_colors('rgb', 1))
```

```python
import random

def shuffle_list(lst):

    shuffled = lst.copy()
    random.shuffle(shuffled)
    return shuffled

print(shuffle_list([1, 2, 3, 4, 5]))

print(shuffle_list(['a', 'b', 'c', 'd']))
```

```python
import random

def shuffle_list(lst):
    shuffled = []
    temp = lst.copy()
    while temp:
        item = random.choice(temp)
        shuffled.append(item)
        temp.remove(item)
    return shuffled
```
