```python
sisters = ('Hafsa', 'Saratu')
brothers = ('Muhd', 'Jabir')
print(sisters)
print(brothers)
```

```python
siblings = sisters + brothers
print(siblings)
```

```python
print(len(siblings))
print(f'I have {len(siblings)} siblings')
```

```python
father = ('Abdullahi',)
mother = ('Safiyya',)

family_members = siblings + father + mother
print(family_members)
print(len(family_members))
```

```python
*siblings_unpacked, father, mother = family_members
print('Siblings:', siblings_unpacked)
print('Father:', father)
print('Mother:', mother)

s1, s2, s3, s4, father, mother = family_members
print(s1, s2, s3, s4, father, mother)
```
