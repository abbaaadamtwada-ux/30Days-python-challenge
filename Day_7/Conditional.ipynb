```python
my_age = 25
your_age = int(input('Enter your age: '))

if your_age > my_age:
    diff = your_age - my_age
    if diff == 1:
        print(f'You are {diff} year older than me.')
    else:
        print(f'You are {diff} years older than me.')
elif your_age < my_age:
    diff = my_age - your_age
    if diff == 1:
        print(f'I am {diff} year older than you.')
    else:
        print(f'I am {diff} years older than you.')
else:
    print('We are the same age!')
```

```python
a = int(input('Enter number one: '))
b = int(input('Enter number two: '))

if a > b:
    print(f'{a} is greater than {b}')
elif a < b:
    print(f'{a} is smaller than {b}')
else:
    print(f'{a} is equal to {b}')
```

```python
score = int(input('Enter your score: '))

if score >= 80 and score <= 100:
    print('Grade: A')
elif score >= 70 and score <= 79:
    print('Grade: B')
elif score >= 60 and score <= 69:
    print('Grade: C')
elif score >= 50 and score <= 59:
    print('Grade: D')
elif score >= 0 and score <= 49:
    print('Grade: F')
else:
    print('Invalid score! Enter 0 to 100')
```

```python
month = input('Enter month: ').capitalize()  # capitalize() makes 'september' -> 'September'

if month in ['September', 'October', 'November']:
    print(f'The season is Autumn.')
elif month in ['December', 'January', 'February']:
    print(f'The season is Winter.')
elif month in ['March', 'April', 'May']:
    print(f'The season is Spring.')
elif month in ['June', 'July', 'August']:
    print(f'The season is Summer.')
else:
    print('Invalid month!')
```

```python
fruits = ['banana', 'orange', 'mango', 'lemon']

fruit = input('Enter a fruit: ').lower()

if fruit in fruits:
    print('That fruit already exists in the list')
else:
    fruits.append(fruit)
    print('Modified list:', fruits)
```

```python
person = {
    'first_name': 'Asabeneh',
    'last_name': 'Yetayeh',
    'age': 250,
    'country': 'Finland',
    'is_marred': True,
    'skills': ['JavaScript', 'React', 'Node', 'MongoDB', 'Python'],
    'address': {
        'street': 'Space street',
        'zipcode': '02210'
    }
}
print(person)
```

```python
if 'skills' in person:
    skills = person['skills']
    middle_index = len(skills) // 2  # 5 // 2 = 2
    print('Middle skill:', skills[middle_index])
```

```python
skills = person['skills']

if skills == ['JavaScript', 'React']:
    print('He is a front end developer')
elif set(skills) == {'Node', 'Python', 'MongoDB'}:
    print('He is a back end developer')
elif set(skills) == {'React', 'Node', 'MongoDB'}:
    print('He is a fullstack developer')
else:
    print('Unknown title')

if 'JavaScript' in skills and 'React' in skills and len(skills) == 2:
    print('He is a front end developer')
elif 'Node' in skills and 'Python' in skills and 'MongoDB' in skills and len(skills) == 3:
    print('He is a back end developer')
elif 'React' in skills and 'Node' in skills and 'MongoDB' in skills and len(skills) == 3:
    print('He is a fullstack developer')
else:
    print('Unknown title - because he has 5 skills, he is more than that!')
```
